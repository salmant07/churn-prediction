```python
import pandas as pd

df = pd.read_csv("../data/BankChurners.csv")
df = df.drop(columns=["CLIENTNUM"] + [c for c in df.columns if c.startswith("Naive_Bayes")])
df["churn"] = (df["Attrition_Flag"] == "Attrited Customer").astype(int)

print(df.shape)
print(df["churn"].mean())
df.head()
```
